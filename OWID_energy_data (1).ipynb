### TIMESERIES FORECASTING (Fb PROPHET MODEL)

```python

import numpy as np
from prophet import Prophet

   ###  Load data
df = pd.read_csv('cleaned_data.csv')

   ### Handling columns
country_col = 'country' if 'country' in df.columns else 'entity'
df = df.rename(columns={'year': 'ds', country_col: 'entity'})
df['ds'] = pd.to_datetime(df['ds'], format='%Y')

target_column = 'low_carbon_share_elec' 
target_goal_percentage = 100.0
forecast_horizon_years = 30

results = []

for country_name in df['entity'].unique():
    cdf = df[df['entity'] == country_name][['ds', target_column]].dropna().copy()
    cdf = cdf.rename(columns={target_column: 'y'})
    
    if len(cdf) < 5:
        continue
        
    cdf['cap'] = 100.0
    m = Prophet(growth='logistic', yearly_seasonality=False, weekly_seasonality=False, daily_seasonality=False)
    m.fit(cdf)
    
    
    future = m.make_future_dataframe(periods=30, freq='YE')
    future['cap'] = 100.0
    forecast = m.predict(future)
    
    
    attained_rows = forecast[forecast['yhat'] >= target_goal_percentage]
    attained_year = int(attained_rows.iloc[0]['ds'].year) if not attained_rows.empty else np.nan

    
    
    start_val = forecast.iloc[0]['yhat']
    end_val = forecast.iloc[-1]['yhat']
    total_growth_pct = ((end_val - start_val) / max(start_val, 1e-5)) * 100
    
    results.append({
        'country': country_name,
        'target_attained_year': attained_year,
        'projected_final_percentage': round(end_val, 2),
        'total_percentage_growth': round(total_growth_pct, 2)
    })

results_df = pd.DataFrame(results)

print(results_df.sort_values(by='target_attained_year', na_position='last'))
```

```python
results_df.to_csv('clean_energy_forecast_summary.csv',index=False)
print("summary results saved sucessfully!")
```

```python
target_year = 2050
target_share = 100.0

results = []

for country_name in ['India', 'China', 'United States']:

    
    # Filtering  data for the country and drop missing values for the metric
    temp_df = df[df['entity'] == country_name][['ds', 'low_carbon_share_elec']].dropna()
    
    
    if not temp_df.empty:
        
        latest_row = temp_df.loc[temp_df['ds'].idxmax()]
        latest_year = latest_row['ds'].year
        latest_share = latest_row['low_carbon_share_elec']
        
        # Calculate time remaining
        years_remaining = target_year - latest_year
        
        if years_remaining > 0 and latest_share < target_share:
            # Absolute growth rate (percentage points needed per year)
            points_per_year = (target_share - latest_share) / years_remaining
            
            # Compound Annual Growth Rate (CAGR)
            cagr = ((target_share / latest_share) ** (1 / years_remaining) - 1) * 100
            
            results.append({
                'Country': country_name,
                'Latest Year': latest_year,
                'Latest Share (%)': round(latest_share, 2),
                'Points Needed/Year': round(points_per_year, 2),
                'CAGR (%)': round(cagr, 2)
            })

# Converting results into a DataFrame 
import pandas as pd
growth_df = pd.DataFrame(results)
print(growth_df.to_string(index=False))
```

```python
import pandas as pd
from prophet import Prophet

all_forecasts = []

for country_name in ['India', 'China', 'United States']:
    
    cdf = df[df['entity'] == country_name].copy()
    cdf = cdf[['ds', 'low_carbon_share_elec']].rename(columns={'low_carbon_share_elec': 'y'})
    cdf = cdf.dropna()
    cdf['cap'] = 100.0
    
    # Fitting Prophet model
    m = Prophet(growth='logistic', yearly_seasonality=False, weekly_seasonality=False, daily_seasonality=False)
    m.fit(cdf)
    
    # 3. Creating  future dataframe up to 2050
    future = m.make_future_dataframe(periods=25, freq='YS')
    future['cap'] = 100.0
    
    # 4. Prediction
    forecast = m.predict(future)
    
    
    expanded_forecast = forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper', 'trend']].copy()
    
    expanded_forecast['entity'] = country_name
    
    expanded_forecast = pd.merge(expanded_forecast, cdf, on='ds', how='left')
    
    all_forecasts.append(expanded_forecast)

# Combine all countries into one master DataFrame for tableau visuaization
master_forecast_df = pd.concat(all_forecasts, ignore_index=True)

# Save to CSV for Tableau
master_forecast_df.to_csv('clean_energy_master_forecast.csv', index=False)
print("Master forecast file successfully created and ready for Tableau!")
```
